### Aggregating data

Now, most csv files include timestamps which stop me from having one row per id, day_in_study, and study_interval. To address that, I will aggregate the data depending on their type for each csv file.

**Features and data type**

*wrist_temperature:*
I am using "temperature_diff_from_baseline" which is a level so the aggregation method will be mean, std, min, and max.

*heart_rate_variability_details:*
I am using "rmssd" and "high_frequency" which are a levels so the aggregation method will be mean, std, min, and max. I inlcuded std because if there's variation in HRV throughout the day, that can be a strong indicative of stress.

*resting_heart_rate:*
I am using "value" but the file has no timestamp column, it's a row per study_day so non-applicable.

*demographic_vo2_max:*
I am using "filtered_demographic_vo2_max" but the file has no timestamp column, it's a row per study_day so non-applicable.

*exercise:*
I am using "activeduration" which is a duration so the aggregation method will be summation.

*hormones_and_selfreport:*
I am using "phase", "estrogen", and "lh" but the file has no timestamp column, it's a row per study_day so non-applicable.

```python
key = ['id', 'study_interval', 'day_in_study']
temperature_daily = (
    temperature.groupby(key)
    .agg(
          mean_temp=("temperature_diff_from_baseline", "mean"),
          max_temp=("temperature_diff_from_baseline", "max"),
          min_temp=("temperature_diff_from_baseline", "min"),
          std_temp=("temperature_diff_from_baseline", "std"),
    )
    .reset_index()
)

print("Daily rows:", len(temperature_daily))
print("Key dupes:", temperature_daily.duplicated(key).sum())
temperature_daily.head()
```

So now what about the days where there was only one row to begin with. Now the std column will have NaN. This will be addressed in the "handling missing data" section.

But why do I even have std? The interesting result isn't "std predicts stress", it's whether it predicts stress more in the luteal phase than the follicular phase. I get at this by looking at feature importance within each phase separately.

****std is a candidate feature. rmssd_std, for instance, measures how much a person's HRV swung across the day, a plausible stress signal that's distinct from the daily average. It goes into your feature matrix, and the feature-importance/SHAP step later to tell me whether it actually predicted stress.****

Also, I am including max and min because for example, a day that's calm except for one sharp temperature spike has a high max but a smooth std and this can be captured and interpreted.

Here is the same code for the other files:

```python
hrv_daily = (
    hrv.groupby(key)
    .agg(
          mean_rmssd=("rmssd", "mean"),
          max_rmssd=("rmssd", "max"),
          min_rmssd=("rmssd", "min"),
          std_rmssd=("rmssd", "std"),

          mean_high_frequency=("high_frequency", "mean"),
          max_high_frequency=("high_frequency", "max"),
          min_high_frequency=("high_frequency", "min"),
          std_high_frequency=("high_frequency", "std"),
    )
    .reset_index()
)

print("Daily rows:", len(hrv_daily))
print("Key dupes:", hrv_daily.duplicated(key).sum())
hrv_daily.head()
```

### Handling Missing Data

Since we ensured data fell within a specific range and aggregated the data, there are no missing values within the _daily files. We can see that only std has NaN values and that's because if std is undefined for a single data point, it measures spread across multiple readings, so a day with only one reading has no spread to compute. These NaNs therefore indicate days where only one measurement was recorded, not missing data in the usual sense.

To deal with that, and because we want to maintain the std column, we fill the NaN values with 0. Filling with 0 reflects no observed spread while preserving the day's other valid statistics (mean, min, max) rather than dropping the row and losing data.

For the rest I will use imputation but I need to first decide whether I will use statistical methods or time-series depending on the gap of missing values. **By gap I mean the typical longest-gap across participant-waves**.

Let's check temperature first:

```python
# this code finds the longest run of consecutive missing days per participant
def max_gap(s):
    return s.isna().astype(int).groupby(s.notna().cumsum()).sum().max()

print(merged.groupby(['id','study_interval'])['mean_temp'].apply(max_gap).describe())
```

For temperature, most gaps are short approx 2 days, but the tail is 32 which is very long meaning some participants have a very long stretch of not having any recordings. For the short gaps, I will use time-series imputation.

I will use interpolation instead of forward-fill because temperature is a signal that changes gradually and continiously and I can capture that through interpolation.

```python
key_sort = ['id', 'study_interval', 'day_in_study']
merged = merged.sort_values(key_sort)

merged['mean_temp'] = (merged
    .groupby(['id', 'study_interval'])['mean_temp']
    .transform(lambda s: s.interpolate(method='linear', limit=3, limit_area='inside')))

# filling the rest with the median

merged['mean_temp'] = merged['mean_temp'].fillna(merged['mean_temp'].median())
```

I will run the same process for the rest of the columns in my data and write out which method I used and why after the code.

```python
cols_to_check = ['max_temp', 'min_temp', 'mean_rmssd', 'min_rmssd', 'max_rmssd', 'mean_high_frequency', 'min_high_frequency', 'max_high_frequency',
                 'mean_bpm', 'min_bpm', 'max_bpm', 'mean_vo2', 'min_vo2', 'max_vo2']

for col in cols_to_check:
    gaps = merged.groupby(['id', 'study_interval'])[col].apply(max_gap)
    print(f"{col}: mean gap = {gaps.mean():.1f}")
```

```python
key_sort = ['id', 'study_interval', 'day_in_study']
merged = merged.sort_values(key_sort)

merged['min_temp'] = (merged
    .groupby(['id', 'study_interval'])['min_temp']
    .transform(lambda s: s.interpolate(method='linear', limit=3, limit_area='inside')))

# filling the rest with the median

merged['min_temp'] = merged['min_temp'].fillna(merged['min_temp'].median())
```

```python
key_sort = ['id', 'study_interval', 'day_in_study']
merged = merged.sort_values(key_sort)

merged['max_temp'] = (merged
    .groupby(['id', 'study_interval'])['max_temp']
    .transform(lambda s: s.interpolate(method='linear', limit=3, limit_area='inside')))

# filling the rest with the median

merged['max_temp'] = merged['max_temp'].fillna(merged['max_temp'].median())
```

```python
median_cols = [
    'mean_vo2', 'max_vo2', 'min_vo2',
    'mean_rmssd', 'max_rmssd', 'min_rmssd',
    'mean_high_frequency', 'max_high_frequency', 'min_high_frequency',
    'mean_bpm', 'max_bpm', 'min_bpm'
]

for col in median_cols:
    merged[col] = merged[col].fillna(merged[col].median())
```

```python
# see if any 'left alone' column has extreme outliers worth capping
for col in ['estrogen', 'lh', 'max_temp', 'min_temp', 'mean_rmssd', 'min_rmssd', 'max_rmssd', 'mean_high_frequency', 'min_high_frequency', 'max_high_frequency',
                 'mean_bpm', 'min_bpm', 'max_bpm', 'mean_vo2', 'min_vo2', 'max_vo2']:
    q1, q3 = merged[col].quantile([0.25, 0.75])
    iqr = q3 - q1
    upper, lower = q3 + 3*iqr, q1 - 3*iqr
    n_out = ((merged[col] > upper) | (merged[col] < lower)).sum()
    print(f"{col}: {n_out} extreme outliers beyond 3×IQR")
```

But looking at the outlier values below, are they actually extreme? No, so I don't want to capp and risk losing valuable information about the relationship between these features and stress.

```python
def show_outlier_values(col):
    q1, q3 = merged[col].quantile([0.25, 0.75])
    iqr = q3 - q1
    upper, lower = q3 + 3*iqr, q1 - 3*iqr
    outliers = merged[(merged[col] > upper) | (merged[col] < lower)][col]
    print(f"{col}: normal range {lower:.1f} to {upper:.1f}")
    print(f"  extreme values: {sorted(outliers.round(1).tolist())}")

for col in ['lh', 'max_temp']:
    show_outlier_values(col)
```

I wanted to check if any other feature needed smoothing and before checking through code, temperature would be the one to check because it shouldn't be varying a lot through for a participant during a study interval. The rest can fluctuate more (except vo2 which doens't vary a lot anyway).

```python
key_sort = ['id', 'study_interval', 'day_in_study']
merged = merged.sort_values(key_sort)

for col in ['estrogen', 'lh', 'mean_temp', 'mean_bpm', 'mean_rmssd', 'mean_vo2']:

    daily_change = (merged.groupby(['id','study_interval'])[col]
                    .diff().abs().mean())
    overall_std = merged[col].std()
    ratio = daily_change / overall_std
    print(f"{col}: avg daily change = {daily_change:.2f}, "
          f"std = {overall_std:.2f}, ratio = {ratio:.2f}")
```

As expected temperature is the one with a high ratio and may need smoothing. I looked at how big is a typical daily jump compared to the temperature signal's whole range.

Wrist temperature should move slowly not jump. Even across menstural cycle phases, it fluctuates, but that's a multi-day fluctuation not spikes day-to-day. But I want to do another check to ensure that these fluctuations are actually noise and not signals I could be ignoring.

```python
key_sort = ['id', 'study_interval', 'day_in_study']
merged = merged.sort_values(key_sort)

# looks at how big are day-to-day temp jumps, relative to the signal's range
daily_change = merged.groupby(['id','study_interval'])['mean_temp'].diff().abs()
print("Median daily change:", daily_change.median())
print("Overall temp std:   ", merged['mean_temp'].std())
print("Ratio:", daily_change.median() / merged['mean_temp'].std())
```

The median day-to-day change relative to the overall spread was 0.41, indicating moderate noise. The mean-based ratio was higher (0.45), and the gap between the two shows that while most daily changes are modest, a few are large. According to Simic et al. (2025) [link text](https://pmc.ncbi.nlm.nih.gov/articles/PMC12608656/#B9-sensors-25-06592), wrist temperature during a stress event may lead to a drop of 1 to 3 degrees C. Therefore, these big jumps could potentially be due to stress and I shouldn't smooth.

### Data Splitting and Validation

**Data Splitting**

I will be doing a **stratified** data split.

**Validation**

I am evaluating the models using 5-fold cross-validation with StratifiedGroupKFold (split data into 5 parts, train on 4, test on 1, rotate 5 times, average the results).
- This groups by participant ID so the same person never appears in both training and test folds — important because there are repeated daily measurements per participant, and mixing them would let the model recognise individuals rather than learn stress patterns.
  - StratifiedGroupKFold does this by assigning whole participants to folds, not individual rows.
- It also stratifies by stress class, preserving the Low/Moderate/High balance in each fold.
  - StratifiedGroupKFold does this where while assigning participants to folds, it also tries to make each fold have roughly the same proportion of Low/Moderate/High as the overall data.

**It's 5-fold so that 4/5 (80%) is for training and 1/5 (20%) is for testing**

```python
import optuna
from optuna.samplers import TPESampler
from sklearn.model_selection import cross_val_score, StratifiedGroupKFold
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score
from sklearn.model_selection import cross_val_predict


optuna.logging.set_verbosity(optuna.logging.WARNING)

X = merged[feature_cols]
y = merged[target_col]
groups = merged[group_col]
sgkf = StratifiedGroupKFold(n_splits=5)

def objective_lr(trial):
    C = trial.suggest_float('C', 1e-3, 100, log=True)
    pipe = Pipeline([
        ('scaler', StandardScaler()),
        ('clf', LogisticRegression(penalty='l2', C=C,
                                   class_weight='balanced', max_iter=1000)),
    ])
    return cross_val_score(pipe, X, y, groups=groups,
                           cv=sgkf, scoring='f1_macro', n_jobs=-1).mean()

study_lr = optuna.create_study(direction='maximize', sampler=TPESampler(seed=42))
study_lr.optimize(objective_lr, n_trials=30)


preds_lr = cross_val_predict(
    Pipeline([('scaler', StandardScaler()),
              ('clf', LogisticRegression(penalty='l2', C=study_lr.best_params['C'],
                                         class_weight='balanced', max_iter=1000))]),
    X, y, groups=groups, cv=sgkf)

print("LogReg — best macro F1:", round(study_lr.best_value, 3))
print("Accuracy:", round(accuracy_score(y, preds_lr), 3))
print("LogReg — best params:", study_lr.best_params)
```

thw pred-lr rebuilds the pipeline using the winning C, then runs the folds again but this time keeping the predictions. Every row gets a prediction from a model that didn't train on it. Now I have an array to score.

**Random Forest**

```python
from sklearn.ensemble import RandomForestClassifier
def objective_rf(trial):
    model = RandomForestClassifier(
        n_estimators     = 150,
        max_depth        = trial.suggest_int('max_depth', 3, 14),
        min_samples_leaf = trial.suggest_int('min_samples_leaf', 2, 30),
        max_features     = trial.suggest_categorical('max_features', ['sqrt', 'log2']),
        class_weight     = 'balanced',
        random_state     = 42,
        n_jobs           = -1,
    )
    return cross_val_score(model, X, y, groups=groups,
                           cv=sgkf, scoring='f1_macro').mean()

study_rf = optuna.create_study(direction='maximize', sampler=TPESampler(seed=42))
study_rf.optimize(objective_rf, n_trials=15)
rf = RandomForestClassifier(**study_rf.best_params, n_estimators=150,
                            class_weight='balanced', random_state=42, n_jobs=-1)
preds_rf = cross_val_predict(rf, X, y, groups=groups, cv=sgkf)

print("Best macro F1:", round(study_rf.best_value, 3))
print("Accuracy:", round(accuracy_score(y, preds_rf), 3))
print("Best params:", study_rf.best_params)
```

**Feature Importance**

Remember I will compute feature importance separately for each phase's model, then compare. **The idea is to use feature importance to answer my research question**. For example "In the luteal phase, temperature ranked highest". This is useful to answer the question.

I will get feature importance through:
- Built-in importance automatically produced by the **tree models** during training.
- For the **logistic regression**, the size of each feature's coefficient tells me its influence (after scaling, so they're comparable). Bigger absolute coefficient = stronger effect on the prediction.

I could also use **permutation importance** for **pruning** (still deciding)


**Interpretability**

To interpret the data I will use **SHAP**. It is a method for explaining why a model made a particular prediction and which features were most important which is perfect. It looks at how much a feature contributed to making this prediction. Importantly, it includes direction, e.g. high temperature pushes toward high stress.

I'm doing feature importance for the three models that did best. This is permutation importance: Take the fitted model, shuffle the values of one feature to create random nonsense, and then evaluate how much the macro F1 score decreases. A significant drop indicates that the model relied on that feature.

**Permutation Importance**

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.inspection import permutation_importance

tr_idx, te_idx = next(sgkf.split(X, y, groups=groups))
X_tr, X_te = X.iloc[tr_idx], X.iloc[te_idx]
y_tr, y_te = y.iloc[tr_idx], y.iloc[te_idx]

def get_importance(model):
    model.fit(X_tr, y_tr)
    r = permutation_importance(model, X_te, y_te, scoring='f1_macro',
                               n_repeats=20, random_state=42, n_jobs=-1)
    return pd.Series(r.importances_mean, index=feature_cols).sort_values()

imp_dt  = get_importance(dt)
imp_rf  = get_importance(rf)
imp_xgb = get_importance(xgb)

def colours(idx):
    out = []
    for f in idx:
        if f in ('estrogen', 'lh'):        out.append('#c98a5e')
        elif 'temp' in f:                  out.append('#7aa874')
        elif 'rmssd' in f or 'freq' in f:  out.append('#6b9ac4')
        else:                              out.append('#a0a0a0')
    return out

fig, axes = plt.subplots(1, 3, figsize=(18, 6), sharex=True)

for ax, imp, name in zip(axes,
                         [imp_dt, imp_rf, imp_xgb],
                         ['Decision Tree', 'Random Forest', 'XGBoost']):
    ax.barh(imp.index, imp.values, color=colours(imp.index), edgecolor='black', lw=0.4)
    ax.axvline(0, c='grey', lw=0.8)
    ax.set_title(f'{name}')
    ax.set_xlabel('Drop in macro F1 when shuffled')

fig.suptitle('Permutation Importance — coloured by signal category', fontweight='bold')
plt.tight_layout()
plt.show()
```

As we can see, not much predictive work is happening here. mean_vo2, mean_temp / std_temp / max_temp, and mean_bpm appear near the top of all three panels.
